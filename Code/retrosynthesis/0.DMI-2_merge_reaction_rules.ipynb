# DMI-01 反应规则合并\n
\n
将 MNX 与 DrugBank 的 reaction atom-mapping rules 合并。\n
缺失字段使用空字符串填充。\n

```python
import pandas as pd

# 输入文件
mnxreac_smile_atom_mapping_rules_file_path = '/home/wuke/project/microbiome–drug/Results/MNXreaction_smiles_atommap_rules_filter_0.2.csv'
drugbank_reaction_new_atommap_rules_file_path = './drugbank_reaction_new_atommap_rules.csv'

# 读取数据
df_mnx = pd.read_csv(mnxreac_smile_atom_mapping_rules_file_path)
df_drugbank = pd.read_csv(drugbank_reaction_new_atommap_rules_file_path)

# 合并：按列名自动对齐，缺失值补空
merged_df = pd.concat([df_mnx, df_drugbank], ignore_index=True, sort=False)
merged_df = merged_df.fillna('')

# 输出文件
output_file = './merged_atommap_rules.csv'
merged_df.to_csv(output_file, index=False)

print('MNX rows:', len(df_mnx))
print('DrugBank rows:', len(df_drugbank))
print('Merged rows:', len(merged_df))
print('Merged columns:', len(merged_df.columns))
print('Output:', output_file)
merged_df
```
