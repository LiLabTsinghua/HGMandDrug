```python
import matplotlib.pyplot as plt

colors = ['#0571b0', '#92c5de', '#f4a582', '#ca0020']
labels = ['#0571b0', '#92c5de', '#f4a582', '#ca0020']

fig, ax = plt.subplots(figsize=(2, 0.6), dpi=200)

for i, (c, lab) in enumerate(zip(colors, labels)):
    # 画色块
    ax.add_patch(plt.Rectangle((i, 0.2), 1, 0.6, color=c))
    
    # 写标签（在色块中间）
    ax.text(i + 0.5, 0.5, lab,
            ha='center', va='center',
            fontsize=5,
            color='white' if i in [0, 3] else 'black')  # 深色用白字更清晰

ax.set_xlim(0, len(colors))
ax.set_ylim(0, 1)
ax.axis('off')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 假设 result_num_df 是一个 Pandas DataFrame，包含 reaction_num 列
import pandas as pd

# 示例数据
all_data = [1274,1274,1274]
result1 = [98,814,944]


# 设置通用参数
plt.figure(figsize=(1.8, 2), dpi=400)
plt.rcParams.update({'font.size': 7})
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['pdf.fonttype'] = 42

# 设置轴样式
plt.xticks([], fontsize=7)  
plt.yticks(fontsize=7)
plt.gca().spines['top'].set_linewidth(0)
plt.gca().spines['bottom'].set_linewidth(1)
plt.gca().spines['left'].set_linewidth(1)
plt.gca().spines['right'].set_linewidth(0)

# 设置柱状图宽度
width = 0.5
colors = ['#92c5de','#0571b0', '#ca0020']
# 创建柱状图
# 计算动态 x 轴坐标
x = np.arange(len(all_data)) + 0.5  # 使 x 坐标从 0.5 开始，间隔 1
bars = plt.bar(x, all_data, width, color='none', edgecolor='black', linewidth=1,zorder=1)

for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 20,  # 20 可根据实际调整位置
             f'{all_data[i]}', ha='center', va='bottom', fontsize=7)
    
x = np.arange(len(result1)) + 0.5  # 使 x 坐标从 0.5 开始，间隔 1
bars = plt.bar(x, result1, width, color=colors, edgecolor='black', linewidth=0,zorder=0)

for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 20,  # 20 可根据实际调整位置
             f'{result1[i]}', ha='center', va='bottom', fontsize=7)
    
# 设置坐标轴参数
plt.tick_params(axis='y', direction='out', width=1, which='both', length=4, pad=2)
plt.tick_params(axis='x', direction='out', width=1, which='both', length=4, pad=1)
plt.ylim(0, 1500)  # 调整 y 轴范围
plt.xlim(0, 3)  # 调整 x 轴范围，使其适应柱子

# 设置 x 轴刻度标签
plt.xticks(x, ['BioTransformer','MicrobeRX','This study'] , rotation=40, fontsize=8,ha='right')

# 设置 y 轴标签
plt.ylabel('Drug-metabolite associations\nfrom DrugBank', fontsize=9, fontweight='bold')

plt.savefig('../../Figure/Figure-1b.pdf', dpi=400, bbox_inches='tight')
# 显示图形
plt.show()
```
